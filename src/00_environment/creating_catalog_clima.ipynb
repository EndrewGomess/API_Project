```python
spark.sql("CREATE CATALOG IF NOT EXISTS clima;")

spark.sql(""" COMMENT ON CATALOG clima IS 'Catalog containing the climate, environmental, and predictive analytics data for the project.' """)
```

```python
spark.sql("USE CATALOG clima;")
```

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS bronze;")
spark.sql(""" COMMENT ON SCHEMA clima.bronze IS 'Bronze layer containing raw data ingested from external data sources.' """)
 
spark.sql("CREATE SCHEMA IF NOT EXISTS silver;")
spark.sql(""" COMMENT ON SCHEMA clima.silver IS 'Silver layer containing cleaned, standardized, and transformed data.' """)

spark.sql("CREATE SCHEMA IF NOT EXISTS gold;")
spark.sql(""" COMMENT ON SCHEMA clima.gold IS 'Gold layer containing curated data prepared for analytics, dashboards, and machine learning.' """)
```

```python
spark.sql("USE SCHEMA bronze;")
```

```python
spark.sql("CREATE VOLUME IF NOT EXISTS air_quality;")
spark.sql(""" COMMENT ON VOLUME clima.bronze.air_quality IS 'Stores raw air quality data retrieved from the Open-Meteo API.' """)

spark.sql("CREATE VOLUME IF NOT EXISTS location_data;")
spark.sql(""" COMMENT ON VOLUME clima.bronze.location_data IS 'Stores raw location and geographic data used by the project.' """)

spark.sql("CREATE VOLUME IF NOT EXISTS nasa_data;")
spark.sql(""" COMMENT ON VOLUME clima.bronze.nasa_data IS 'Stores raw data retrieved from NASA data sources used by the project.' """)

spark.sql("CREATE VOLUME IF NOT EXISTS weather_data;")
spark.sql(""" COMMENT ON VOLUME clima.bronze.weather_data IS 'Stores raw weather data retrieved from the Open-Meteo API.' """)
```
